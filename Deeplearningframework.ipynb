**Q12.** What is the basic structure of a PyTorch neural network ?

ANS: A PyTorch neural network is generally created using the torch.nn.Module class. The basic structure consists of input layer → hidden layers → output layer, along with a forward pass that defines how data moves through the network.

1. Import PyTorch

```python
import torch
import torch.nn as nn
```

2. Define the Neural Network

```python
class NeuralNetwork(nn.Module):

    def __init__(self):
        super(NeuralNetwork, self).__init__()

        self.layer1 = nn.Linear(10, 64)
        self.layer2 = nn.Linear(64, 32)
        self.output = nn.Linear(32, 1)

    def forward(self, x):
        x = torch.relu(self.layer1(x))
        x = torch.relu(self.layer2(x))
        x = self.output(x)
        return x
```

3. Create the Model

```python
model = NeuralNetwork()
```

Structure :

Input Layer → Hidden Layer 1 → Hidden Layer 2 → Output Layer

10 inputs

   ↓

Linear Layer (64 neurons)

   ↓

ReLU

   ↓

Linear Layer (32 neurons)

   ↓

ReLU

   ↓

Output Layer (1 neuron)

Important Components:

-> nn.Module → Base class for PyTorch neural networks.

-> __init__() → Defines the layers of the network.

-> nn.Linear() → Creates a fully connected layer.

-> forward() → Defines how input data flows through the layers.

-> Activation functions such as ReLU → Introduce non-linearity.

**Q16.** What are some common activation functions used in neural networks ?

ANS : Common Activation Functions Used in Neural Networks
Activation functions decide whether and how strongly a neuron should be activated. They introduce non-linearity, allowing neural networks to learn complex patterns.

1. ReLU — Rectified Linear Unit

Formula: f(x) = max(0, x)

-> Outputs 0 for negative values.

-> Outputs the input for positive values.

-> Very commonly used in hidden layers.

-> Computationally fast.

```python
nn.ReLU()
```

2. Sigmoid

Formula: f(x) = 1 / (1 + e⁻ˣ)

-> Produces values between 0 and 1.

-> Commonly used for binary classification output layers.

-> Can suffer from the vanishing gradient problem.

```python
nn.Sigmoid()
```

3. Tanh — Hyperbolic Tangent

-> Range: -1 to +1

-> Produces both positive and negative values.

-> Historically common in hidden layers.

-> Can also suffer from vanishing gradients.

```python
nn.Tanh()
```

4. Softmax

-> Converts outputs into probabilities.

-> The probabilities sum to 1.

-> Commonly used in the output layer for multi-class classification.

```python
nn.Softmax(dim=1)
```

5. Leaky ReLU

-> Similar to ReLU but allows a small negative output instead of making all negative values zero.

-> Helps reduce the dying ReLU problem.

Common activation functions used in neural networks include ReLU, Sigmoid, Tanh, Softmax, and Leaky ReLU. ReLU is widely used in hidden layers, Sigmoid is commonly used for binary classification, Tanh produces values between -1 and 1, Softmax is used for multi-class classification, and Leaky ReLU helps overcome the dying ReLU problem.

**Q8.**How do you implement a custom loss function in PyTorch ?


```python
import torch
import torch.nn as nn
import torch.optim as optim

# Custom Mean Absolute Error loss
def custom_loss(predictions, targets):
    loss = torch.mean(torch.abs(predictions - targets))
    return loss


# Create a simple model
model = nn.Linear(4, 1)

# Define optimizer
optimizer = optim.Adam(model.parameters(), lr=0.001)

# Sample input and target
x = torch.randn(5, 4)
y = torch.randn(5, 1)

# Forward pass
predictions = model(x)

# Calculate custom loss
loss = custom_loss(predictions, y)

# Backpropagation
optimizer.zero_grad()
loss.backward()
optimizer.step()

print("Custom Loss:", loss.item())
```

```python
class CustomLoss(nn.Module):

    def __init__(self):
        super(CustomLoss, self).__init__()

    def forward(self, predictions, targets):
        return torch.mean(torch.abs(predictions - targets))


loss_function = CustomLoss()

predictions = model(x)
loss = loss_function(predictions, y)

print("Custom Loss:", loss.item())
```
